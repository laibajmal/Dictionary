🎟️ TASK 6: EVENT ATTENDEES
━━━━━━━━━━━━━━━━━━

An event has a list of people who registered. Some people registered more than once.

Use a set to find the total number of unique attendees.

Expected Output:

Unique Attendees: 6

```python
registrations = ["Maryam", "Areeba", "Fahad", "Laiba", "Areeba", "Hadia", "Sara", "Maryam"]
print("Unique Attendees: ", len(set(registrations)))
```

Create a dictionary containing a student's name, age, and marks.

Use a for-in loop to display the information.

Expected Output:

Name: Ali
Age: 21
Marks: 85

```python
student = {
    "Name" : "Laiba Ajmal",
    "Age" : 21,
    "Marks" : 85
    }
for key in student:
  print(f"{key} = {student[key]}")
```

🛒 TASK 8: PRODUCT PRICES
━━━━━━━━━━━━━━━━━━

Create a dictionary containing 4 products and their prices.

Use a for-in loop to display each product and its price.

Expected Output:

Milk: 250
Bread: 150
Eggs: 300
Juice: 200

```python
products = {
    "Milk" : 250,
    "Bread" : 150,
    "Egg" : 300,
    "Juice" : 200
}

for key in products:
  print(f"{key}: {products[key]}")
```

📱 TASK 9: MOBILE PHONE
━━━━━━━━━━━━━━━━━━

Create a dictionary containing information about a mobile phone:

* Brand
* Model
* Price
* Storage

Display all the information using a for-in loop.

Expected Output:

Brand: Samsung
Model: Galaxy S24
Price: 180000
Storage: 256GB

━━━━━━━━━━━━━━━━━━

```python
mobile_info = {
    "Brand": "Samsung",
    "Model": "Galaxy S24",
    "Price": 180000,
    "Storage": "256GB"
}
for key in mobile_info:
  print(f"{key}: {mobile_info[key]}")
```
